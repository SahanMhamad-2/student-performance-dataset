```python
def sample_mean_calculator(input_dist,sample_size,n_samples):
    sample_means = []
    for i in range(n_samples):
        sample=np.random.choice(input_dist, size=sample_size,replace=False)
        sample_means.append(sample.mean())
    return sample_means
```

```python
sns.displot(sample_mean_calculator(x, 10, 10), color='green', kde=True)
_ = plt.xlim(-1.5, 1.5)
```

## 1. Mean and Standard Deviation
The **Mean** is the arithmetic average, and **Standard Deviation** measures the dispersion of the data.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Sample Dataset
data = np.array([85, 90, 78, 92, 88, 76, 95, 89, 82, 80])

mean_val = np.mean(data)
std_val = np.std(data)

print(f'Mean: {mean_val:.2f}')
print(f'Standard Deviation: {std_val:.2f}')
```

2. Z-Scores and Raw Scores (X)

```python
# Z-scores for all data points
z_scores = stats.zscore(data)
print(f'Z-scores: {z_scores}')

# Finding X from a Z-score (e.g., Z = 1.2)
z_input = 1.2
x_val = mean_val + (z_input * std_val)
print(f'Raw score (X) for Z=1.2: {x_val:.2f}')
```

3. Normal vs. Skewed Distributions

```python
normal_dist = np.random.normal(50, 10, 1000)
skewed_dist = stats.skewnorm.rvs(a=10, loc=50, scale=10, size=1000)

plt.figure(figsize=(10, 5))
plt.hist(normal_dist, bins=30, alpha=0.5, label='Normal Distribution', color='skyblue')
plt.hist(skewed_dist, bins=30, alpha=0.5, label='Right Skewed', color='salmon')
plt.title("Visualizing Skewness")
plt.legend()
plt.show()
```
